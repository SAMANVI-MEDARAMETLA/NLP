```python
# EX-1: Regular Expressions in Python to Match and Search Patterns in Text

import re

text = "My email is student@gmail.com and my phone number is 9876543210."

email = re.search(r'\S+@\S+', text)
phone = re.search(r'\d{10}', text)

print("Email:", email.group() if email else "Not found")
print("Phone:", phone.group() if phone else "Not found")

print("All numbers:", re.findall(r'\d+', text))
```

```python
# EX-2: Finite State Automaton to Recognize Strings Ending with 'ab'

def check_string(s):
    state = 0

    for ch in s:
        if state == 0:
            state = 1 if ch == 'a' else 0
        elif state == 1:
            state = 1 if ch == 'a' else 2
        elif state == 2:
            state = 1 if ch == 'a' else 0

    return state == 2

s = input("Enter a string: ")

if check_string(s):
    print("Accepted")
else:
    print("Rejected")
```

```python
# EX-3: Morphological Analysis Using NLTK Library

import nltk
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

ps = PorterStemmer()

text = "The students are playing and studying happily."

words = word_tokenize(text)

print("Morphological Analysis:")

for word in words:
    print(word, "->", ps.stem(word))
```

```python
# EX-4: Finite State Machine for Morphological Parsing

def plural_form(word):
    if word.endswith(('s', 'x', 'ch', 'sh')):
        return word + "es"
    elif word.endswith('y') and word[-2] not in 'aeiou':
        return word[:-1] + "ies"
    else:
        return word + "s"

word = input("Enter a noun: ")

print("Singular:", word)
print("Plural:", plural_form(word))
```

```python
# EX-6: N-Gram Model for Text Generation Using Bigram Model

import random
import nltk
from nltk import bigrams, word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

text = "Natural language processing is interesting. "
text += "Natural language processing helps computers understand language."

words = word_tokenize(text.lower())

bigram_list = list(bigrams(words))

model = {}

for w1, w2 in bigram_list:
    model.setdefault(w1, []).append(w2)

word = "natural"
result = [word]

for i in range(10):
    if word not in model:
        break
    word = random.choice(model[word])
    result.append(word)

print("Generated Text:")
print(" ".join(result))
```

```python
# EX-7: Part-of-Speech Tagging Using NLTK Library

import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

text = "The intelligent student writes Python programs."

words = word_tokenize(text)

tags = nltk.pos_tag(words)

print("POS Tagging:")

for word, tag in tags:
    print(word, ":", tag)
```

```python
# EX-8: Stochastic Part-of-Speech Tagging Using a Probabilistic Model

import nltk

nltk.download('brown')
nltk.download('universal_tagset')

from nltk.corpus import brown
from nltk.tag import UnigramTagger

train_data = brown.tagged_sents(tagset='universal')

tagger = UnigramTagger(train_data)

sentence = ["The", "dog", "runs", "quickly"]

result = tagger.tag(sentence)

print("Stochastic POS Tagging:")

for word, tag in result:
    print(word, ":", tag)
```

```python
# EX-9: Rule-Based Part-of-Speech Tagging Using Regular Expressions

import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

patterns = [
    (r'.*ing$', 'VBG'),
    (r'.*ed$', 'VBD'),
    (r'.*ly$', 'RB'),
    (r'.*ous$', 'JJ'),
    (r'.*s$', 'NNS'),
    (r'.*', 'NN')
]

text = "The students are playing happily"

words = word_tokenize(text)

tagger = nltk.RegexpTagger(patterns)

print(tagger.tag(words))
```

```python
# EX-10: Transformation-Based Tagging Using Transformation Rules

import nltk

from nltk.tag import brill, brill_trainer
from nltk.tag import UnigramTagger
from nltk.corpus import treebank

nltk.download('treebank')

train_data = treebank.tagged_sents()[:3000]

initial_tagger = UnigramTagger(train_data)

templates = brill.brill24()

trainer = brill_trainer.BrillTaggerTrainer(
    initial_tagger, templates
)

tagger = trainer.train(train_data, max_rules=10)

sentence = ["The", "student", "is", "learning"]

print("Transformation-Based Tagging:")
print(tagger.tag(sentence))
```

```python
# EX-11: Top-Down Parser for Context-Free Grammars

import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'boy' | 'ball'
V -> 'kicked'
""")

parser = nltk.RecursiveDescentParser(grammar)

sentence = ['the', 'boy', 'kicked', 'the', 'ball']

print("Top-Down Parsing:")

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```

```python
# EX-12: Earley Parser for Context-Free Grammars

import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'cat' | 'mouse'
V -> 'chased'
""")

parser = nltk.EarleyChartParser(grammar)

sentence = ['the', 'cat', 'chased', 'the', 'mouse']

print("Earley Parsing:")

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```

```python
# EX-13: Generate a Parse Tree Using Context-Free Grammar

import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'student' | 'book'
V -> 'reads'
""")

parser = nltk.ChartParser(grammar)

sentence = ['the', 'student', 'reads', 'the', 'book']

print("Parse Tree:")

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```

```python
# EX-14: Check Agreement in Sentences Using Context-Free Grammar Rules

import nltk

grammar = nltk.CFG.fromstring("""
S -> NP_SG VP_SG | NP_PL VP_PL
NP_SG -> 'he' | 'she'
NP_PL -> 'they' | 'we'
VP_SG -> 'runs' | 'plays'
VP_PL -> 'run' | 'play'
""")

parser = nltk.ChartParser(grammar)

sentences = [
    ['he', 'runs'],
    ['they', 'run'],
    ['he', 'run'],
    ['they', 'runs']
]

for sentence in sentences:
    result = list(parser.parse(sentence))

    if result:
        print(" ".join(sentence), "-> Correct Agreement")
    else:
        print(" ".join(sentence), "-> Incorrect Agreement")
```

```python
# EX-15: Probabilistic Context-Free Grammar Parsing

import nltk

grammar = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.8] | 'John' [0.2]
VP -> V NP [0.7] | V [0.3]
Det -> 'the' [1.0]
N -> 'dog' [0.5] | 'cat' [0.5]
V -> 'sees' [1.0]
""")

parser = nltk.ViterbiParser(grammar)

sentence = ['John', 'sees', 'the', 'dog']

print("PCFG Parsing:")

for tree in parser.parse(sentence):
    print(tree)
    print("Probability:", tree.prob())
```

```python
# EX-17: Access WordNet to Retrieve Synsets and Explore Word Meanings

import nltk
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')

word = input("Enter a word: ")

synsets = wordnet.synsets(word)

print("Word Meanings:")

for syn in synsets:
    print("\nSynset:", syn.name())
    print("Definition:", syn.definition())
    print("Examples:", syn.examples())

print("\nSynonyms:")

for syn in synsets:
    for lemma in syn.lemmas():
        print(lemma.name())
```

```python
# EX-18: Simple FOPC Parser for Basic Logical Expressions

import re

def parse_fopc(expression):
    expression = expression.strip()

    pattern = r'([A-Za-z]+)\(([^()]*)\)'

    match = re.fullmatch(pattern, expression)

    if match:
        predicate = match.group(1)
        arguments = match.group(2).split(',')

        print("Predicate:", predicate)
        print("Arguments:")

        for arg in arguments:
            print(arg.strip())

    else:
        print("Invalid FOPC Expression")

expression = "Student(John)"

parse_fopc(expression)
```

```python
# EX-19: Word Sense Disambiguation Using Lesk Algorithm

import nltk
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt')
nltk.download('punkt_tab')

sentence = "He deposited money in the bank"

words = word_tokenize(sentence)

sense = lesk(words, 'bank')

print("Sentence:", sentence)

if sense:
    print("Selected Sense:", sense.name())
    print("Meaning:", sense.definition())
else:
    print("No meaning found")
```

```python
# EX-20: Information Retrieval System Using TF-IDF

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Natural language processing deals with human language.",
    "Machine learning helps computers learn from data.",
    "Python is a programming language.",
    "NLP is used in text processing and analysis."
]

query = input("Enter your search query: ")

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

query_vector = vectorizer.transform([query])

scores = cosine_similarity(query_vector, tfidf_matrix)

print("\nSearch Results:")

for index in scores[0].argsort()[::-1]:
    print(documents[index], "-> Score:", round(scores[0][index], 3))
```

```python
# EX-21: Syntax-Driven Semantic Analysis by Extracting Noun Phrases and Their Meanings

import nltk
from nltk import RegexpParser, pos_tag, word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

text = "The intelligent student reads a book."

words = word_tokenize(text)

tagged_words = pos_tag(words)

grammar = "NP: {<DT>?<JJ>*<NN.*>+}"

parser = RegexpParser(grammar)

tree = parser.parse(tagged_words)

print("Syntax-Driven Semantic Analysis:")

for subtree in tree.subtrees():
    if subtree.label() == 'NP':
        phrase = " ".join(word for word, tag in subtree.leaves())
        print("Noun Phrase:", phrase)

        print("Meaning: A person, place, thing, or concept")
```

```python
# EX-22: Reference Resolution Within a Text

import re

text = "John is a student. He studies Python. He likes programming."

sentences = re.split(r'(?<=[.!?])\s+', text)

previous_name = None

print("Reference Resolution:")

for sentence in sentences:
    words = sentence.split()

    for word in words:
        clean_word = word.strip(".,!?")

        if clean_word.lower() in ['he', 'she', 'they']:
            if previous_name:
                print(clean_word, "refers to", previous_name)

        elif clean_word.istitle():
            previous_name = clean_word

    print(sentence)
```

```python
# EX-23: Evaluate the Coherence of a Given Text

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text = [
    "Natural language processing is a branch of AI.",
    "NLP helps computers understand human language.",
    "It is used in chatbots and translation systems."
]

vectorizer = TfidfVectorizer()

matrix = vectorizer.fit_transform(text)

similarity = cosine_similarity(matrix)

print("Sentence Similarity Matrix:")
print(similarity)

score = 0

for i in range(len(text) - 1):
    score += similarity[i][i + 1]

coherence = score / (len(text) - 1)

print("\nCoherence Score:", round(coherence, 3))

if coherence > 0.1:
    print("Text has some semantic coherence")
else:
    print("Text has low semantic coherence")
```

```python
# EX-24: Recognize Dialog Acts in a Given Dialog or Conversation

def dialog_act(sentence):
    sentence = sentence.lower()

    if sentence.startswith(('hi', 'hello', 'hey')):
        return "Greeting"

    elif sentence.endswith('?'):
        return "Question"

    elif sentence.startswith(('please', 'kindly')):
        return "Request"

    elif sentence.startswith(('thank',)):
        return "Thanking"

    elif sentence.startswith(('bye', 'goodbye')):
        return "Goodbye"

    else:
        return "Statement"

dialog = [
    "Hello!",
    "How are you?",
    "Please help me.",
    "Thank you.",
    "Goodbye!"
]

print("Dialog Act Recognition:")

for sentence in dialog:
    print(sentence, "->", dialog_act(sentence))
```

```python
# EX-26: Machine Translation Using Hugging Face Transformers

from transformers import MarianMTModel, MarianTokenizer

model_name = "Helsinki-NLP/opus-mt-en-fr"

tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

text = input("Enter English text: ")

tokens = tokenizer(text, return_tensors="pt", padding=True)

translated = model.generate(**tokens)

result = tokenizer.decode(translated[0], skip_special_tokens=True)

print("\nEnglish Text:", text)
print("French Translation:", result)
```

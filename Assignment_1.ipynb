```python

# TASK 2: REGULAR EXPRESSION PATTERNS

import re

print("=" * 60)
print("HOSPITAL RECORD - REGULAR EXPRESSION SYSTEM")
print("=" * 60)

# Sample hospital record
text = """
Patient ID: P1234
Date: 12-10-2026
Phone: 9876543210
Email: patient@gmail.com
Age: 25
Medicine: Paracetamol
"""

# Regular expression patterns
patterns = {
    "Patient ID": r"P\d{4}",
    "Date": r"\d{2}-\d{2}-\d{4}",
    "Phone": r"\b\d{10}\b",
    "Email": r"[\w.-]+@[\w.-]+\.\w+",
    "Age": r"\b\d{1,3}\b",
    "Medicine": r"Paracetamol"
}

# Extract information
print("\nEXTRACTED HOSPITAL INFORMATION\n")

for label, pattern in patterns.items():
    matches = re.findall(pattern, text, re.IGNORECASE)
    print(label, ":", matches)

# Test validation
print("\nPATTERN VALIDATION")

patient_id = "P1234"

if re.fullmatch(r"P\d{4}", patient_id):
    print(patient_id, "is a valid Patient ID")
else:
    print(patient_id, "is an invalid Patient ID")

print("\nRESULT")
print("Regular expressions successfully extracted")
print("important information from hospital records.")
print("Task 2 completed successfully.")
```

```python

# TASK 3: TOKENISATION AND TEXT NORMALISATION

import re

print("=" * 60)
print("TEXT TOKENISATION AND NORMALISATION")
print("=" * 60)

text = "PATIENTS are suffering from FEVER and taking MEDICINES."

# Step 1: Convert to lowercase
normalized = text.lower()

# Step 2: Remove punctuation
normalized = re.sub(r"[^\w\s]", "", normalized)

# Step 3: Tokenisation
tokens = normalized.split()

# Step 4: Abbreviation expansion
abbreviations = {
    "bp": "blood pressure",
    "hr": "heart rate",
    "temp": "temperature",
    "fever": "high body temperature"
}

print("\nOriginal Text:")
print(text)

print("\nNormalized Text:")
print(normalized)

print("\nTokens:")
print(tokens)

print("\nExpanded Medical Terms:")
for word in tokens:
    print(word, ":", abbreviations.get(word, word))

print("\nRESULT")
print("Text tokenisation and normalisation completed successfully.")
```

```python

# TASK 4: DFA, NFA AND EQUIVALENCE

import re

print("=" * 60)
print("FINITE STATE AUTOMATA")
print("=" * 60)

# DFA for Patient ID: P followed by 4 to 6 digits
def patient_dfa(value):
    if not value.startswith("P"):
        return False

    digits = value[1:]

    if 4 <= len(digits) <= 6 and digits.isdigit():
        return True

    return False

# NFA for words starting with 'para'
def para_nfa(word):
    states = {0}

    for char in word:
        new_states = set()

        for state in states:
            if state < 4:
                if char == "para"[state]:
                    new_states.add(state + 1)

        if 4 in states:
            new_states.add(4)

        states = new_states

    return 4 in states

# Test DFA
print("\nDFA TESTING")

ids = ["P1234", "P12345", "P123", "A1234"]

for value in ids:
    print(value, ":", "Accepted" if patient_dfa(value) else "Rejected")

# Test NFA
print("\nNFA TESTING")

words = ["para", "paracetamol", "paracetamol500", "fever"]

for word in words:
    print(word, ":", "Accepted" if para_nfa(word) else "Rejected")

# Regular expression equivalence
print("\nREGEX EQUIVALENCE")

pattern = r"P\d{4,6}"

for value in ids:
    regex_result = bool(re.fullmatch(pattern, value))
    dfa_result = patient_dfa(value)

    print(value, ":", "Equivalent" if regex_result == dfa_result else "Not Equivalent")

print("\nRESULT")
print("DFA, NFA and regular expression tests completed successfully.")
```

```python

# TASK 6: DERIVATIONAL MORPHOLOGY

print("=" * 60)
print("DERIVATIONAL MORPHOLOGY ANALYSIS")
print("=" * 60)

# Prefix and suffix examples
derivations = {
    "unhealthy": ("un", "healthy", "prefix"),
    "treatment": ("", "treat", "suffix: ment"),
    "careless": ("", "care", "suffix: less"),
    "recheck": ("re", "check", "prefix"),
    "helpful": ("", "help", "suffix: ful"),
    "discomfort": ("dis", "comfort", "prefix")
}

print("\nDERIVATIONAL ANALYSIS\n")

for word, details in derivations.items():
    prefix, root, affix = details

    print("Word:", word)
    print("Root Word:", root)
    print("Affix:", prefix if prefix else affix)
    print("Type:", affix if prefix else "Suffix")
    print("-" * 40)

# Derivational word formation
def add_prefix(word, prefix):
    return prefix + word

def add_suffix(word, suffix):
    return word + suffix

print("\nWORD FORMATION")

print("healthy ->", add_prefix("healthy", "un"))
print("treat ->", add_suffix("treat", "ment"))
print("care ->", add_suffix("care", "less"))

print("\nRESULT")
print("Derivational morphology analysis completed successfully.")
```

```python

# TASK 7: FINITE-STATE MORPHOLOGICAL PARSING

print("=" * 60)
print("FINITE STATE MORPHOLOGICAL PARSER")
print("=" * 60)

# Morphological dictionary
lexicon = {
    "patients": ("patient", "Noun + Plural"),
    "medicines": ("medicine", "Noun + Plural"),
    "treated": ("treat", "Verb + Past"),
    "treating": ("treat", "Verb + Present Participle"),
    "records": ("record", "Noun + Plural"),
    "diagnoses": ("diagnosis", "Noun + Plural")
}

# Morphological parser
def morphological_parser(word):
    if word in lexicon:
        root, feature = lexicon[word]
        return root, feature

    return None, "Word not found"

# Test words
test_words = [
    "patients",
    "medicines",
    "treated",
    "treating",
    "records",
    "diagnoses",
    "hospital"
]

print("\nPARSING RESULTS\n")

for word in test_words:
    root, feature = morphological_parser(word)

    print("Input Word:", word)
    print("Root:", root)
    print("Morphological Feature:", feature)
    print("-" * 40)

print("\nRESULT")
print("Finite-state morphological parsing demonstration completed.")
```

```python

# TASK 8: ORTHOGRAPHIC RULES AND TWO-LEVEL MORPHOLOGY

print("=" * 60)
print("ORTHOGRAPHIC RULES AND TWO-LEVEL MORPHOLOGY")
print("=" * 60)

# Orthographic plural formation
def plural_form(word):
    irregular = {
        "diagnosis": "diagnoses",
        "analysis": "analyses"
    }

    if word in irregular:
        return irregular[word]

    if word.endswith("y") and len(word) > 1:
        if word[-2] not in "aeiou":
            return word[:-1] + "ies"

    if word.endswith(("s", "x", "ch", "sh")):
        return word + "es"

    return word + "s"

# Orthographic past tense
def past_tense(word):
    if word.endswith("e"):
        return word + "d"

    if word.endswith("y") and len(word) > 1:
        if word[-2] not in "aeiou":
            return word[:-1] + "ied"

    return word + "ed"

print("\nPLURAL FORMATION")

for word in ["patient", "medicine", "diagnosis", "study"]:
    print(word, "->", plural_form(word))

print("\nPAST TENSE FORMATION")

for word in ["treat", "check", "care", "study"]:
    print(word, "->", past_tense(word))

# Two-level morphology examples
mappings = {
    "patient + PL": "patients",
    "medicine + PL": "medicines",
    "treat + PAST": "treated",
    "care + PAST": "cared"
}

print("\nTWO-LEVEL MORPHOLOGY")

for lexical, surface in mappings.items():
    print(lexical, "->", surface)

print("\nRESULT")
print("Orthographic and two-level morphology rules demonstrated.")
```

```python

# TASK 10: PYTHON IMPLEMENTATION AND TESTING

import re

print("=" * 60)
print("HOSPITAL TEXT NORMALISATION SYSTEM")
print("=" * 60)

# Hospital records
records = [
    "Patient P1234 has fever and is taking paracetamol.",
    "Patient P2345 has headache and needs treatment.",
    "Patient P3456 has cough and is taking medicine.",
    "Patient P4567 has fever and body pain.",
    "Patient P5678 has diabetes and requires checkup."
]

# Normalisation function
def normalize_text(text):
    text = text.lower()
    text = re.sub(r"[^\w\s]", "", text)
    return text

# Tokenisation function
def tokenize(text):
    return normalize_text(text).split()

# Search function
def search_records(keyword):
    keyword = normalize_text(keyword)

    results = []

    for record in records:
        if keyword in normalize_text(record):
            results.append(record)

    return results

# Display records
print("\nHOSPITAL RECORDS\n")

for record in records:
    print(record)

# Test normalisation
print("\nNORMALISATION TEST")

sample = "PATIENTS are taking MEDICINES!"
print("Original:", sample)
print("Normalized:", normalize_text(sample))

# Test tokenisation
print("\nTOKENISATION TEST")
print(tokenize(sample))

# Test searching
print("\nSEARCH TEST")

query = "fever"
results = search_records(query)

print("Search Keyword:", query)
print("Matching Records:")

for result in results:
    print("-", result)

print("\nRESULT")
print("Hospital text normalisation and search system tested successfully.")
```

```python

# TASK 12: OCR AND REGIONAL-LANGUAGE CODE-MIXING

import re

print("=" * 65)
print("OCR AND CODE-MIXING TEXT NORMALISATION")
print("=" * 65)

# Sample OCR-generated text
ocr_text = "Pat1ent has fev3r and needs medic1ne."

# OCR correction dictionary
ocr_corrections = {
    "pat1ent": "patient",
    "fev3r": "fever",
    "medic1ne": "medicine"
}

# OCR correction
def correct_ocr(text):
    words = text.split()
    corrected = []

    for word in words:
        clean_word = re.sub(r"[^\w]", "", word.lower())

        if clean_word in ocr_corrections:
            corrected.append(ocr_corrections[clean_word])
        else:
            corrected.append(word)

    return " ".join(corrected)

print("\nOCR PROCESSING")

print("Original OCR Text:")
print(ocr_text)

corrected_text = correct_ocr(ocr_text)

print("\nCorrected Text:")
print(corrected_text)

# Regional-language code-mixing example
code_mix = {
    "jwaram": "fever",
    "mandhu": "medicine",
    "rogam": "disease",
    "noppulu": "pain"
}

def translate_code_mix(text):
    words = text.lower().split()
    translated = []

    for word in words:
        translated.append(code_mix.get(word, word))

    return " ".join(translated)

print("\nCODE-MIXING PROCESS")

mixed_text = "patient ki jwaram undi mandhu kavali"

print("Original Text:")
print(mixed_text)

translated_text = translate_code_mix(mixed_text)

print("\nNormalised Text:")
print(translated_text)

# Final output
print("\nRESULT")
print("OCR correction and regional-language code-mixing")
print("normalisation demonstrated successfully.")
```

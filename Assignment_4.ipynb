```python

# Task 1: Meaning Representation in NLP
# Government Pension Helpdesk

# Knowledge Base
employees = {
    "Ravi": {"service_years": 15},
    "Rahul": {"service_years": 8},
    "Priya": {"service_years": 12}
}

MINIMUM_SERVICE = 10

# 1. Answering a Question
def check_eligibility(name):
    if name in employees:
        years = employees[name]["service_years"]

        if years >= MINIMUM_SERVICE:
            return name + " satisfies the service condition."
        else:
            return name + " does not satisfy the service condition."

    return "Employee details not found."


# 2. Checking a Claim
def verify_claim(name, claimed_years):
    if name not in employees:
        return "Claim cannot be verified."

    actual_years = employees[name]["service_years"]

    if actual_years == claimed_years:
        return "Claim is TRUE."

    return "Claim is FALSE."


# 3. Drawing an Inference
def inference(name):
    if name in employees:
        years = employees[name]["service_years"]

        if years >= MINIMUM_SERVICE:
            return name + " satisfies the qualifying service rule."

    return "The service condition is not satisfied."


# 4. Matching Differently Worded Enquiries
def match_enquiry(question):
    question = question.lower()

    if ("eligible" in question or "qualify" in question
            or "receive" in question or "get" in question):
        if "pension" in question:
            return "Meaning: Check pension eligibility."

    return "Meaning not identified."


# Program Execution
print("GOVERNMENT PENSION HELPDESK")
print("--------------------------------")

print("\n1. Answering a Question")
print(check_eligibility("Ravi"))

print("\n2. Checking a Claim")
print(verify_claim("Ravi", 15))

print("\n3. Drawing an Inference")
print(inference("Priya"))

print("\n4. Matching Different Enquiries")
q1 = "Is Ravi eligible for pension?"
q2 = "Can Ravi receive a pension?"

print("Question 1:", match_enquiry(q1))
print("Question 2:", match_enquiry(q2))
```

```python

# Task 2: Meaning Structure of Language
# Government Pension Helpdesk

# Knowledge Base
employees = {
    "Ravi": {"service": 15, "eligible": True},
    "Rahul": {"service": 8, "eligible": False}
}

MINIMUM_SERVICE = 10

# 1. Verifiability
def verify_statement(name):
    if name not in employees:
        return "UNKNOWN"

    if employees[name]["service"] >= MINIMUM_SERVICE:
        return "TRUE"
    else:
        return "FALSE"


# 2. Unambiguous Representation
def identify_meaning(question):
    question = question.lower()

    if "every employee" in question:
        return "Universal Quantifier"

    elif "some employee" in question:
        return "Existential Quantifier"

    return "Meaning identified"


# 3. Canonical Form
def canonical_form(question):
    question = question.lower()

    if "eligible" in question or "receive" in question:
        if "pension" in question:
            return "EligibleForPension(Ravi)"

    return "Unknown Representation"


# 4. Inference
def inference(name):
    if name in employees:
        service = employees[name]["service"]

        if service >= MINIMUM_SERVICE:
            return name + " satisfies the service condition."
        else:
            return name + " does not satisfy the service condition."

    return "Insufficient Information"


# 5. Expressiveness
def check_uncertainty(statement):
    if "think" in statement.lower() or "may" in statement.lower():
        return "Uncertain Statement"
    return "Certain Statement"


# Program Execution
print("GOVERNMENT PENSION HELPDESK")
print("--------------------------------")

print("\n1. VERIFIABILITY")
print("Ravi:", verify_statement("Ravi"))
print("Rahul:", verify_statement("Rahul"))
print("Meena:", verify_statement("Meena"))

print("\n2. UNAMBIGUOUS REPRESENTATION")
print(identify_meaning("Every employee gets pension"))
print(identify_meaning("Some employee gets pension"))

print("\n3. CANONICAL FORM")
print(canonical_form("Is Ravi eligible for pension?"))
print(canonical_form("Can Ravi receive a pension?"))

print("\n4. INFERENCE")
print(inference("Ravi"))
print(inference("Rahul"))

print("\n5. EXPRESSIVENESS")
print(check_uncertainty("I think I may be eligible"))
print(check_uncertainty("I am eligible for pension"))
```

```python

# Task 3: First Order Predicate Calculus (FOPC)
# Government Pension Helpdesk

# Knowledge Base
employees = {
    "Ravi": {"service": 15, "retired": True},
    "Rahul": {"service": 8, "retired": True},
    "Priya": {"service": 12, "retired": False}
}

dependents = {
    "Lakshmi": "Ravi"
}

pensioners = ["Ravi"]
family_pensioners = ["Lakshmi"]

MIN_SERVICE = 10


# 1. Predicates
def Employee(name):
    return name in employees

def Pensioner(name):
    return name in pensioners

def Eligible(name):
    return Employee(name) and employees[name]["service"] >= MIN_SERVICE

def Dependent(name):
    return name in dependents

def Retired(name):
    return Employee(name) and employees[name]["retired"]

def FamilyPensioner(name):
    return name in family_pensioners


# 2. Universal Quantifier
print("1. UNIVERSAL QUANTIFIER")
for name in employees:
    if Employee(name) and Eligible(name):
        print(name, "satisfies service condition")


# 3. Existential Quantifier
print("\n2. EXISTENTIAL QUANTIFIER")
found = any(Eligible(name) for name in employees)
print("Some employee satisfies service condition:", found)


# 4. Negation
print("\n3. NEGATION")
print("Rahul is not eligible:", not Eligible("Rahul"))


# 5. Exactly One Relationship
print("\n4. EXACTLY ONE RELATIONSHIP")
for person, employee in dependents.items():
    print(person, "is dependent of", employee)


# 6. Model Evaluation
print("\n5. MODEL EVALUATION")
print("Employee(Ravi):", Employee("Ravi"))
print("Pensioner(Ravi):", Pensioner("Ravi"))
print("Eligible(Rahul):", Eligible("Rahul"))
print("Employee(Priya):", Employee("Priya"))
print("Dependent(Ravi):", Dependent("Ravi"))
print("FamilyPensioner(Lakshmi):",
      FamilyPensioner("Lakshmi"))


# 7. Inference
print("\n6. INFERENCE")

for name in employees:
    if Employee(name) and Eligible(name):
        print(name, "satisfies the pension service condition.")

# Family pension inference
for dependent, employee in dependents.items():
    if Employee(employee):
        print(dependent,
              "is a family pension dependent of", employee)


# 8. Recursive Ancestor Relationship
print("\n7. RECURSIVE RELATIONSHIP")

parents = {
    "Ravi": "Ramesh",
    "Ramesh": "Suresh"
}

def Ancestor(person, descendant):
    if descendant not in parents:
        return False

    parent = parents[descendant]

    if parent == person:
        return True

    return Ancestor(person, parent)

print("Ravi is ancestor of Suresh:",
      Ancestor("Ravi", "Suresh"))
```

```python

# Task 5: Syntax-Driven Semantic Analysis
# Government Pension Helpdesk

# 1. Grammar Rules
grammar = {
    "S": "NP + VP",
    "NP": "ProperNoun / Determiner + Noun",
    "VP": "Verb + NP",
    "Noun": "Employee / Pensioner / Pension",
    "Verb": "Receives / Approves / Submits",
    "Determiner": "Every / Some"
}

print("1. GRAMMAR RULES")
for rule, value in grammar.items():
    print(rule, "->", value)


# 2. Lambda Expressions
def employee(x):
    return f"Employee({x})"

def pensioner(x):
    return f"Pensioner({x})"

def receives(x, y):
    return f"Receives({x},{y})"

def approves(x, y):
    return f"Approves({x},{y})"


# 3. Lambda Reduction
print("\n2. LAMBDA REDUCTION")

expression = employee("Ravi")
print("Lambda Expression: λx.Employee(x)")
print("Reduced Expression:", expression)


# 4. Semantic Analysis
def semantic_analysis(sentence):
    words = sentence.lower().replace(".", "").replace("?", "").split()

    if "ravi" in words and "employee" in words:
        return employee("Ravi")

    if "ravi" in words and "pension" in words:
        return receives("Ravi", "Pension")

    if "approves" in words and "application" in words:
        return approves("Officer", "Application")

    if "every" in words and "employee" in words:
        return "FOR ALL x: Employee(x) -> Receives(x,Pension)"

    if "some" in words and "employee" in words:
        return "EXISTS x: Employee(x) AND Pensioner(x)"

    return "Meaning not identified"


# 5. Parse Tree Generation
def parse_tree(sentence):
    words = sentence.replace(".", "").replace("?", "").split()

    print("\nSentence:", sentence)
    print("          S")
    print("         / \\")
    print("        NP  VP")

    if len(words) >= 2:
        print("        |    |")
        print("       " + words[0] + "   " + " ".join(words[1:]))


# 6. Deriving Logical Forms
print("\n3. SEMANTIC ANALYSIS")

sentences = [
    "Ravi is an employee.",
    "Ravi receives pension.",
    "Every employee receives pension.",
    "Some employee is pensioner."
]

for sentence in sentences:
    parse_tree(sentence)
    print("Logical Form:", semantic_analysis(sentence))


# 7. Quantifier Handling
print("\n4. QUANTIFIER HANDLING")

every = "FOR ALL x: Employee(x) -> Eligible(x)"
some = "EXISTS x: Employee(x) AND Eligible(x)"

print("Every:", every)
print("Some:", some)


# 8. Coverage Calculation
print("\n5. SEMANTIC ANALYSER COVERAGE")

total = 200
successful = 170
failed = total - successful

coverage = (successful / total) * 100

print("Total Enquiries:", total)
print("Successfully Analysed:", successful)
print("Failed Enquiries:", failed)
print("Coverage:", coverage, "%")
```

```python

# Task 7: Word Sense Disambiguation (WSD)
# Government Pension Helpdesk

import re
from collections import Counter
from sklearn.feature_extraction import DictVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.pipeline import make_pipeline

# 1. Sense Inventory
sense_inventory = {
    "service": {
        "employment": {
            "gloss": "employment work qualifying years retirement employee",
            "related": "job career government staff tenure"
        },
        "customer": {
            "gloss": "customer assistance support help department",
            "related": "service desk enquiry complaint information"
        }
    },
    "benefit": {
        "pension": {
            "gloss": "pension financial support retirement payment",
            "related": "employee family income allowance"
        },
        "advantage": {
            "gloss": "advantage useful facility employment privilege",
            "related": "reward opportunity improvement"
        }
    },
    "dependent": {
        "family": {
            "gloss": "family member financially supported by employee",
            "related": "spouse child parent pension"
        },
        "variable": {
            "gloss": "variable value depends on another variable",
            "related": "mathematics statistics equation"
        }
    }
}

stop_words = {
    "the", "is", "a", "an", "of", "in", "to",
    "and", "for", "my", "how", "does"
}

def tokenize(text):
    return set(re.findall(r"\b[a-z]+\b", text.lower()))


# 2. Basic Simplified Lesk
def basic_lesk(sentence, word):
    context = tokenize(sentence)

    best_sense = None
    maximum_overlap = -1

    print("\nSentence:", sentence)

    for sense, details in sense_inventory[word].items():
        gloss_words = tokenize(details["gloss"])
        overlap = context & gloss_words

        print(sense, "Overlap:", sorted(overlap),
              "Count:", len(overlap))

        if len(overlap) > maximum_overlap:
            maximum_overlap = len(overlap)
            best_sense = sense

    return best_sense


# 3. Lesk with Stop-word Removal
def lesk_without_stopwords(sentence, word):
    context = tokenize(sentence) - stop_words

    scores = {}

    for sense, details in sense_inventory[word].items():
        gloss = tokenize(details["gloss"]) - stop_words
        scores[sense] = len(context & gloss)

    return max(scores, key=scores.get)


# 4. Extended Lesk
def extended_lesk(sentence, word):
    context = tokenize(sentence) - stop_words

    scores = {}

    for sense, details in sense_inventory[word].items():
        gloss = tokenize(details["gloss"])
        related = tokenize(details["related"])

        signature = gloss | related
        scores[sense] = len(context & signature)

    return max(scores, key=scores.get)


# 5. Test Three Lesk Methods
print("WORD SENSE DISAMBIGUATION")
print("--------------------------------")

tests = [
    ("Ravi completed qualifying years of government service",
     "service"),
    ("The department provides customer assistance and support",
     "service"),
    ("The employee receives pension financial support",
     "benefit"),
    ("A spouse is financially supported by the employee",
     "dependent")
]

for sentence, word in tests:
    print("\nWord:", word)
    print("Basic Lesk:", basic_lesk(sentence, word))
    print("Stop-word Lesk:",
          lesk_without_stopwords(sentence, word))
    print("Extended Lesk:", extended_lesk(sentence, word))


# 6. Annotated Training Dataset
# Illustrative examples; expand to 600 occurrences
dataset = [
    ("employee qualifying service years retirement", "employment"),
    ("government service completed years", "employment"),
    ("employee job career tenure", "employment"),
    ("retirement service duration staff", "employment"),
    ("qualifying years employment record", "employment"),
    ("government employee completed service", "employment"),
    ("customer service desk support", "customer"),
    ("department provides customer assistance", "customer"),
    ("help desk enquiry complaint", "customer"),
    ("customer information support department", "customer"),
    ("service desk helps customers", "customer"),
    ("assistance for pension enquiries", "customer"),

    ("pension financial support retirement", "pension"),
    ("employee receives pension payment", "pension"),
    ("family pension income allowance", "pension"),
    ("retirement financial benefit", "pension"),
    ("pension support for retired employees", "pension"),
    ("monthly pension payment", "pension"),
    ("employment advantage privilege facility", "advantage"),
    ("useful advantage opportunity", "advantage"),
    ("employee receives additional facilities", "advantage"),
    ("employment reward and improvement", "advantage"),
    ("job privilege and opportunity", "advantage"),
    ("additional employment advantage", "advantage"),

    ("spouse child family pension dependent", "family"),
    ("employee supports family financially", "family"),
    ("dependent parent receives family pension", "family"),
    ("family member supported by employee", "family"),
    ("spouse financially dependent on employee", "family"),
    ("child receives family pension", "family"),
    ("dependent variable mathematical equation", "variable"),
    ("variable value depends on another variable", "variable"),
    ("statistics dependent variable calculation", "variable"),
    ("mathematics equation variable value", "variable"),
    ("statistical variable changes with another", "variable"),
    ("dependent variable in mathematics", "variable")
]


# 7. Contextual Feature Extraction
def extract_features(sentence):
    words = sentence.lower().split()

    return {
        "words": " ".join(words),
        "first_word": words[0] if words else "",
        "last_word": words[-1] if words else "",
        "has_pension": "pension" in words,
        "has_employee": "employee" in words,
        "has_service": "service" in words,
        "has_family": "family" in words,
        "has_customer": "customer" in words
    }


X = [extract_features(text) for text, label in dataset]
y = [label for text, label in dataset]


# 8. Supervised Classification
print("\n\nSUPERVISED CLASSIFICATION")

model = make_pipeline(
    DictVectorizer(),
    MultinomialNB()
)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scores = cross_val_score(
    model, X, y, cv=cv, scoring="accuracy"
)

print("Five-Fold Accuracy:", scores)
print("Average Accuracy:", round(scores.mean() * 100, 2), "%")

model.fit(X, y)

new_sentence = "employee completed qualifying service years"
prediction = model.predict([extract_features(new_sentence)])

print("New Sentence:", new_sentence)
print("Predicted Sense:", prediction[0])


# 9. Most-Frequent-Sense Baseline
print("\nMOST FREQUENT SENSE BASELINE")

frequency = Counter(y)
most_frequent = frequency.most_common(1)[0]

print("Most Frequent Sense:", most_frequent[0])
print("Frequency:", most_frequent[1])


# 10. Downstream Retrieval Demonstration
print("\nRETRIEVAL WITH AND WITHOUT WSD")

documents = {
    "Circular 101": "employee qualifying service retirement years",
    "Circular 102": "customer service department assistance",
    "Circular 103": "pension financial support retirement payment",
    "Circular 104": "family dependent spouse child pension",
    "Circular 105": "employment benefits and advantages"
}

query = "government service for pension"

print("Original Query:", query)

print("\nWithout WSD:")
for doc, content in documents.items():
    overlap = len(tokenize(query) & tokenize(content))
    if overlap > 0:
        print(doc, "Score:", overlap)

selected_sense = extended_lesk(
    query, "service"
)

expanded_query = query + " " + sense_inventory[
    "service"
][selected_sense]["gloss"]

print("\nSelected Sense:", selected_sense)
print("Expanded Query:", expanded_query)

print("\nWith WSD:")
for doc, content in documents.items():
    overlap = len(tokenize(expanded_query) & tokenize(content))
    if overlap > 0:
        print(doc, "Score:", overlap)
```

```python

# Task 8: Information Retrieval over Pension Circulars

import re
import math
from collections import Counter, defaultdict
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Sample Pension Circulars
documents = {
    "Circular 101": "Pension eligibility requires qualifying service of employees.",
    "Circular 102": "Family pension benefits are provided to eligible dependents.",
    "Circular 103": "Retirement pension depends on employee service records.",
    "Circular 104": "Pension revision increases retirement benefits.",
    "Circular 105": "Government employees receive pension after retirement.",
    "Circular 106": "Qualifying service determines pension eligibility."
}

# 2. Text Preprocessing
stop_words = {
    "the", "is", "are", "to", "of", "and",
    "a", "an", "after", "on", "in"
}

def preprocess(text):
    words = re.findall(r"\b[a-z]+\b", text.lower())
    return [word for word in words if word not in stop_words]

print("1. TEXT PREPROCESSING")

for name, content in documents.items():
    print(name, ":", preprocess(content))


# 3. Term-Document Matrix
print("\n2. TERM-DOCUMENT MATRIX")

corpus = [" ".join(preprocess(text))
          for text in documents.values()]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(corpus)

terms = vectorizer.get_feature_names_out()

print("Terms:", list(terms))
print("Matrix:")

for name, row in zip(documents.keys(), tfidf_matrix.toarray()):
    print(name, ":", row.round(3))


# 4. Inverted Index
print("\n3. INVERTED INDEX")

inverted_index = defaultdict(list)

for name, content in documents.items():
    for word in set(preprocess(content)):
        inverted_index[word].append(name)

for word, doc_list in sorted(inverted_index.items()):
    print(word, ":", doc_list)


# 5. TF-IDF Calculation
print("\n4. TF-IDF CALCULATION")

N = len(documents)

for term in ["pension", "service", "employee",
             "family", "retirement", "eligibility"]:

    df = sum(
        term in preprocess(content)
        for content in documents.values()
    )

    idf = math.log(N / df) if df else 0

    print(
        term,
        "| DF:", df,
        "| IDF:", round(idf, 3)
    )


# 6. Cosine Similarity and Ranking
print("\n5. DOCUMENT RANKING")

queries = [
    "pension qualifying service",
    "family pension dependent",
    "retirement pension benefits",
    "pension revision",
    "employee eligibility",
    "government pension"
]

for query in queries:
    query_vector = vectorizer.transform(
        [" ".join(preprocess(query))]
    )

    scores = cosine_similarity(
        query_vector, tfidf_matrix
    )[0]

    ranking = sorted(
        zip(documents.keys(), scores),
        key=lambda x: x[1],
        reverse=True
    )

    print("\nQuery:", query)

    for name, score in ranking[:3]:
        print(name, "Similarity:", round(score, 3))


# 7. Evaluation Metrics
print("\n6. RETRIEVAL EVALUATION")

def evaluate(relevant, retrieved):
    relevant = set(relevant)
    retrieved = set(retrieved)

    tp = len(relevant & retrieved)
    fp = len(retrieved - relevant)
    fn = len(relevant - retrieved)

    precision = tp / (tp + fp) if tp + fp else 0
    recall = tp / (tp + fn) if tp + fn else 0

    f1 = (
        2 * precision * recall / (precision + recall)
        if precision + recall else 0
    )

    return precision, recall, f1


# Example relevance judgments for one query
relevant_docs = ["Circular 101", "Circular 106"]
retrieved_docs = ["Circular 101", "Circular 106", "Circular 103"]

precision, recall, f1 = evaluate(
    relevant_docs, retrieved_docs
)

print("Precision:", round(precision, 3))
print("Recall:", round(recall, 3))
print("F1 Score:", round(f1, 3))


# 8. Precision at 10
def precision_at_10(relevant, ranked):
    relevant = set(relevant)
    top_10 = ranked[:10]

    return sum(doc in relevant for doc in top_10) / 10


ranked_documents = [
    "Circular 101",
    "Circular 106",
    "Circular 103"
]

p10 = precision_at_10(
    relevant_docs, ranked_documents
)

print("Precision at 10:", round(p10, 3))


# 9. Mean Average Precision
def average_precision(relevant, ranked):
    relevant = set(relevant)

    if not relevant:
        return 0

    hits = 0
    total_precision = 0

    for i, doc in enumerate(ranked, start=1):
        if doc in relevant:
            hits += 1
            total_precision += hits / i

    return total_precision / len(relevant)


ap = average_precision(
    relevant_docs, ranked_documents
)

print("Mean Average Precision:", round(ap, 3))


# 10. Query Expansion Using WSD
print("\n7. INFORMATION RETRIEVAL WITH WSD")

original_query = "pension service"

expanded_query = (
    "pension service employment qualifying "
    "years retirement employee"
)

print("Original Query:", original_query)
print("Expanded Query:", expanded_query)

for query in [original_query, expanded_query]:

    query_vector = vectorizer.transform(
        [" ".join(preprocess(query))]
    )

    scores = cosine_similarity(
        query_vector, tfidf_matrix
    )[0]

    ranking = sorted(
        zip(documents.keys(), scores),
        key=lambda x: x[1],
        reverse=True
    )

    print("\nQuery:", query)

    for name, score in ranking[:3]:
        print(name, "Similarity:", round(score, 3))
```

```python

import re
import time
from collections import Counter, defaultdict

import nltk
from nltk.tokenize import wordpunct_tokenize

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


# ============================================================
# 1. SAMPLE PENSION CIRCULAR CORPUS
# ============================================================

circulars = {
    "C1": "Employees completing ten years of qualifying service are eligible for pension.",
    "C2": "Family pension is provided to eligible dependents after the death of an employee.",
    "C3": "Pension commutation allows a pensioner to receive a lump sum amount.",
    "C4": "Government service rules explain employee retirement and qualifying service.",
    "C5": "Pension revision increases the amount of pension according to revised rules.",
    "C6": "A dependent family member may receive family pension under government rules."
}

queries = [
    "Who qualifies for pension after ten years of service?",
    "Who receives family pension?",
    "What is pension commutation?",
    "What are government service rules?",
    "How is pension revision calculated?",
    "Can a dependent receive family pension?"
]


# ============================================================
# 2. TEXT PREPROCESSING
# ============================================================

stop_words = {
    "the", "is", "are", "a", "an", "of", "to",
    "for", "and", "who", "what", "how", "can",
    "after", "under", "with"
}


def preprocess(text):
    tokens = wordpunct_tokenize(text.lower())

    tokens = [
        word for word in tokens
        if word.isalpha() and word not in stop_words
    ]

    return tokens


# ============================================================
# 3. SEMANTIC GRAMMAR AND LAMBDA-STYLE REDUCTION
# ============================================================

semantic_rules = {
    "employee": "Employee(x)",
    "pensioner": "Pensioner(x)",
    "dependent": "Dependent(x)",
    "receives pension": "ReceivesPension(x)",
    "qualifying service": "QualifyingService(x)",
    "family pension": "FamilyPension(x)"
}


def semantic_analysis(sentence):

    sentence = sentence.lower()

    if "ten years" in sentence and "service" in sentence:
        return "Employee(x) AND QualifyingService(x, 10)"

    elif "family pension" in sentence:
        return "FamilyPension(x)"

    elif "commutation" in sentence:
        return "PensionCommutation(x)"

    elif "pension revision" in sentence:
        return "PensionRevision(x)"

    elif "service rules" in sentence:
        return "GovernmentServiceRules(x)"

    else:
        return "Meaning not identified"


def lambda_reducer(expression, variable, replacement):
    return expression.replace(variable, replacement)


print("\n========== SEMANTIC ANALYSIS ==========")

semantic_start = time.time()

for query in queries:
    logical_form = semantic_analysis(query)
    print("\nEnquiry:", query)
    print("Logical Form:", logical_form)

example_expression = "ReceivesPension(x)"
reduced_expression = lambda_reducer(
    example_expression, "x", "Ravi"
)

print("\nLambda-style Reduction:")
print("Before:", example_expression)
print("After:", reduced_expression)

semantic_time = time.time() - semantic_start


# ============================================================
# 4. FOPC KNOWLEDGE BASE AND INFERENCE
# ============================================================

employees = {
    "Ravi": 15,
    "Rahul": 8,
    "Priya": 12,
    "Lakshmi": 5
}

pensioners = {"Ravi", "Priya"}

dependents = {
    "Ravi": ["Anu"],
    "Priya": ["Kiran"]
}


def qualifying_service(person):
    return employees.get(person, 0) >= 10


def eligible_for_pension(person):
    return qualifying_service(person)


def receives_family_pension(person):
    return person in pensioners


def has_dependent(person):
    return len(dependents.get(person, [])) > 0


print("\n========== FOPC KNOWLEDGE BASE ==========")

fopc_start = time.time()

for person in employees:
    print("\nPerson:", person)
    print("Qualifying Service:", qualifying_service(person))
    print("Pension Eligibility:", eligible_for_pension(person))
    print("Has Dependent:", has_dependent(person))

# Inference example
print("\nInference:")
if eligible_for_pension("Ravi"):
    print("Ravi completed ten years of service.")
    print("Therefore, Ravi is eligible for pension.")

fopc_time = time.time() - fopc_start


# ============================================================
# 5. WORD SENSE DISAMBIGUATION USING LESK
# ============================================================

sense_inventory = {
    "service": {
        "employment": "employee work employment years retirement",
        "public_service": "government public department assistance"
    },

    "benefit": {
        "pension_benefit": "pension payment financial amount",
        "general_benefit": "advantage help support"
    },

    "dependent": {
        "family_member": "family person child spouse pension",
        "dependent_variable": "variable value depends on another"
    }
}


def lesk(sentence, word):

    sentence_words = set(preprocess(sentence))
    best_sense = None
    maximum_overlap = -1

    for sense, gloss in sense_inventory[word].items():

        gloss_words = set(preprocess(gloss))
        overlap = len(sentence_words.intersection(gloss_words))

        if overlap > maximum_overlap:
            maximum_overlap = overlap
            best_sense = sense

    return best_sense, maximum_overlap


print("\n========== WORD SENSE DISAMBIGUATION ==========")

wsd_start = time.time()

wsd_examples = [
    ("The employee completed ten years of service", "service"),
    ("The government provides public service", "service"),
    ("The pension benefit is paid monthly", "benefit"),
    ("The dependent receives family pension", "dependent")
]

for sentence, word in wsd_examples:

    sense, overlap = lesk(sentence, word)

    print("\nSentence:", sentence)
    print("Word:", word)
    print("Selected Sense:", sense)
    print("Gloss Overlap:", overlap)

wsd_time = time.time() - wsd_start


# ============================================================
# 6. INFORMATION RETRIEVAL USING TF-IDF
# ============================================================

print("\n========== INFORMATION RETRIEVAL ==========")

retrieval_start = time.time()

document_ids = list(circulars.keys())
documents = list(circulars.values())

vectorizer = TfidfVectorizer(
    tokenizer=preprocess,
    token_pattern=None
)

document_matrix = vectorizer.fit_transform(documents)

# Display vocabulary
print("\nVocabulary:")
print(vectorizer.get_feature_names_out())

# Inverted index
inverted_index = defaultdict(list)

for index, document in enumerate(documents):

    for term in set(preprocess(document)):
        inverted_index[term].append(document_ids[index])

print("\nInverted Index:")

for term, ids in inverted_index.items():
    print(term, ":", ids)


# Retrieve documents using cosine similarity
def retrieve_documents(query):

    query_vector = vectorizer.transform([query])

    scores = cosine_similarity(
        query_vector,
        document_matrix
    ).flatten()

    ranking = scores.argsort()[::-1]

    return [
        (document_ids[i], round(scores[i], 3))
        for i in ranking
        if scores[i] > 0
    ]


for query in queries:

    print("\nQuery:", query)

    results = retrieve_documents(query)

    if results:
        for document_id, score in results:
            print(document_id, "Similarity:", score)
    else:
        print("No matching circular found.")

retrieval_time = time.time() - retrieval_start


# ============================================================
# 7. TEST CASES FROM PREVIOUS TASKS
# ============================================================

print("\n========== TESTING PREVIOUS TASKS ==========")

# Task 3: FOPC model evaluation
assert qualifying_service("Ravi") == True
assert qualifying_service("Rahul") == False

print("Task 3 FOPC tests: PASSED")

# Task 5: Semantic analysis
assert semantic_analysis(
    "Who receives family pension?"
) == "FamilyPension(x)"

assert semantic_analysis(
    "What is pension commutation?"
) == "PensionCommutation(x)"

print("Task 5 semantic analysis tests: PASSED")

# Task 7: WSD
sense, overlap = lesk(
    "The pension benefit is paid monthly",
    "benefit"
)

assert sense == "pension_benefit"

print("Task 7 WSD test: PASSED")

# Task 8: TF-IDF retrieval
test_results = retrieve_documents(
    "family pension dependent"
)

assert len(test_results) > 0

print("Task 8 retrieval test: PASSED")


# ============================================================
# 8. PIPELINE PERFORMANCE MEASUREMENTS
# ============================================================

print("\n========== PIPELINE PERFORMANCE ==========")

print("Stage                  Processed      Run Time")

print(
    "Semantic Analysis      ",
    len(queries), "enquiries     ",
    round(semantic_time, 5), "seconds"
)

print(
    "FOPC Reasoning         ",
    len(employees), "persons       ",
    round(fopc_time, 5), "seconds"
)

print(
    "WSD                    ",
    len(wsd_examples), "examples      ",
    round(wsd_time, 5), "seconds"
)

print(
    "Information Retrieval ",
    len(documents), "documents     ",
    round(retrieval_time, 5), "seconds"
)


# ============================================================
# 9. RETRIEVAL QUALITY VS VOCABULARY SIZE
# ============================================================

print("\n========== PLOT 1: RETRIEVAL QUALITY ==========")

vocabulary_sizes = [5, 10, 15, 20]
retrieval_quality = []

for size in vocabulary_sizes:

    limited_vectorizer = TfidfVectorizer(
        tokenizer=preprocess,
        token_pattern=None,
        max_features=size
    )

    limited_matrix = limited_vectorizer.fit_transform(documents)

    correct = 0

    for query, expected_id in zip(
        queries,
        ["C1", "C2", "C3", "C4", "C5", "C6"]
    ):

        query_vector = limited_vectorizer.transform([query])

        scores = cosine_similarity(
            query_vector,
            limited_matrix
        ).flatten()

        predicted_id = document_ids[scores.argmax()]

        if predicted_id == expected_id:
            correct += 1

    accuracy = correct / len(queries) * 100
    retrieval_quality.append(accuracy)

    print("Vocabulary:", size, "Accuracy:", round(accuracy, 2), "%")


print("\nRetrieval Quality Plot")

for size, accuracy in zip(
    vocabulary_sizes,
    retrieval_quality
):
    bars = "#" * int(accuracy / 5)
    print(str(size).ljust(5), "|", bars, round(accuracy, 2), "%")


# ============================================================
# 10. WSD ACCURACY VS ANNOTATED DATA SIZE
# ============================================================

print("\n========== PLOT 2: WSD ACCURACY ==========")

annotation_sizes = [20, 40, 60, 80, 100]

# Illustrative values for demonstrating the plot.
# Replace these with measured accuracy from an annotated dataset.
wsd_accuracy = [60, 68, 74, 82, 88]

for size, accuracy in zip(
    annotation_sizes,
    wsd_accuracy
):

    bars = "#" * int(accuracy / 5)

    print(
        str(size).ljust(5),
        "|",
        bars,
        accuracy,
        "%"
    )


# ============================================================
# 11. FINAL RESULT
# ============================================================

print("\n========== FINAL RESULT ==========")

print("Semantic Analysis: Completed")
print("FOPC Reasoning: Completed")
print("Word Sense Disambiguation: Completed")
print("Information Retrieval: Completed")
print("Previous Task Tests: Completed")
print("Performance Measurements: Completed")
print("Comparison Plots: Generated")

print("\nNLP Pension Helpdesk Pipeline Executed Successfully.")
```

```python

from datetime import date
import time

# =====================================================
# 1. PENSION EMPLOYEE DATA
# =====================================================

employees = {
    "Ravi": {
        "retirement_date": date(2019, 6, 30),
        "total_service": 20,
        "service_breaks": 2,
        "break_years": [1, 2],
        "family_member": "Anu"
    },

    "Rahul": {
        "retirement_date": date(2021, 5, 31),
        "total_service": 15,
        "service_breaks": 1,
        "break_years": [2],
        "family_member": "Kiran"
    },

    "Priya": {
        "retirement_date": date(2018, 3, 31),
        "total_service": 12,
        "service_breaks": 0,
        "break_years": [],
        "family_member": "Meena"
    }
}


# =====================================================
# 2. PENSION SCHEME RULES WITH EFFECTIVE DATES
# =====================================================

schemes = [
    {
        "name": "Old Pension Scheme",
        "start": date(2000, 1, 1),
        "end": date(2020, 12, 31),
        "minimum_service": 10
    },

    {
        "name": "Revised Pension Scheme",
        "start": date(2021, 1, 1),
        "end": date(2099, 12, 31),
        "minimum_service": 12
    }
]


# =====================================================
# 3. ORIGINAL SYSTEM: BASIC ELIGIBILITY
# =====================================================

def original_eligibility(person):

    employee = employees[person]

    return employee["total_service"] >= 10


# =====================================================
# 4. EFFECTIVE-DATE RULE SELECTION
# =====================================================

def get_applicable_scheme(retirement_date):

    for scheme in schemes:

        if (
            scheme["start"] <= retirement_date
            <= scheme["end"]
        ):
            return scheme

    return None


# =====================================================
# 5. QUALIFYING SERVICE CALCULATION
# =====================================================

def calculate_qualifying_service(person):

    employee = employees[person]

    total_service = employee["total_service"]

    excluded_breaks = sum(employee["break_years"])

    qualifying_years = total_service - excluded_breaks

    return qualifying_years


# =====================================================
# 6. INTERDEPENDENT RULE PROCESSING
# =====================================================

def check_service_condition(person):

    qualifying_years = calculate_qualifying_service(person)

    return qualifying_years >= 10


def check_pension_condition(person):

    return check_service_condition(person)


def check_family_pension_condition(person):

    employee = employees[person]

    service_eligible = check_pension_condition(person)

    family_member_exists = employee["family_member"] is not None

    return service_eligible and family_member_exists


# =====================================================
# 7. MODIFIED SYSTEM: COMPLETE ELIGIBILITY CHECK
# =====================================================

def modified_eligibility(person):

    employee = employees[person]

    retirement_date = employee["retirement_date"]

    scheme = get_applicable_scheme(retirement_date)

    if scheme is None:
        return False, "No applicable pension scheme", 0

    qualifying_years = calculate_qualifying_service(person)

    minimum_years = scheme["minimum_service"]

    eligible = (
        qualifying_years >= minimum_years
        and check_family_pension_condition(person)
    )

    return eligible, scheme["name"], qualifying_years


# =====================================================
# 8. TEST THREE NEW ENQUIRIES
# =====================================================

print("\n========== TASK 12: NEW ENQUIRY TESTS ==========")

start_time = time.time()

# Enquiry 1: Cross-reference between rules
print("\n1. CROSS-REFERENCE BETWEEN RULES")

person = "Ravi"

print("Enquiry: Is Ravi eligible for revised family pension?")
print("Service condition:", check_service_condition(person))
print("Family pension condition:",
      check_family_pension_condition(person))

# Enquiry 2: Effective-date requirement
print("\n2. EFFECTIVE-DATE REQUIREMENT")

person = "Ravi"

scheme = get_applicable_scheme(
    employees[person]["retirement_date"]
)

print("Enquiry: Which scheme applies to Ravi?")
print("Retirement Date:",
      employees[person]["retirement_date"])

if scheme:
    print("Applicable Scheme:", scheme["name"])
else:
    print("No applicable scheme found.")

# Enquiry 3: Aggregation requirement
print("\n3. SERVICE BREAK AND AGGREGATION")

person = "Rahul"

print("Enquiry: Does Rahul qualify after service breaks?")
print("Total Service:",
      employees[person]["total_service"])

print("Service Breaks:",
      employees[person]["service_breaks"])

print("Qualifying Service:",
      calculate_qualifying_service(person))


# =====================================================
# 9. COMPARE ORIGINAL AND MODIFIED SYSTEMS
# =====================================================

print("\n========== SYSTEM COMPARISON ==========")

for person in employees:

    old_result = original_eligibility(person)

    new_result, scheme_name, years = modified_eligibility(person)

    print("\nEmployee:", person)
    print("Original System Eligibility:", old_result)
    print("Modified System Eligibility:", new_result)
    print("Applicable Scheme:", scheme_name)
    print("Qualifying Years:", years)


# =====================================================
# 10. PERFORMANCE MEASUREMENT
# =====================================================

end_time = time.time()

print("\n========== PERFORMANCE MEASUREMENT ==========")

print("Employees Processed:", len(employees))
print("Schemes Available:", len(schemes))
print("New Enquiries Tested:", 3)

print(
    "Execution Time:",
    round(end_time - start_time, 6),
    "seconds"
)


# =====================================================
# 11. FINAL DESIGN RECOMMENDATION
# =====================================================

print("\n========== FINAL DESIGN RECOMMENDATION ==========")

print("Decision: MODIFY THE EXISTING SYSTEM")

print("\nImprovements Added:")
print("1. Interdependent pension rule processing")
print("2. Effective-date based scheme selection")
print("3. Service break calculation")
print("4. Qualifying service aggregation")
print("5. Threshold-based eligibility checking")

print("\nNLP Pension Helpdesk Adaptation Completed.")
```
